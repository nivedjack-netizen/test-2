```python
df = df.drop_duplicates()
```

```python
df["Category"] = df["Category"].fillna("Unknown")
df["Region"] = df["Region"].fillna("Unknown")
df["Customer_Segment"] = df["Customer_Segment"].fillna("Unknown")
```

```python
print("Average Sales:", df["Sales"].mean())
```

```python
category = df.groupby("Category")["Sales"].sum()
print(category)
```

```python
category = df.groupby("Category")["Sales"].sum()
print(category)
```

```python
print("Highest Category:", category.idxmax())
```

```python
print("Lowest Category:", category.idxmin())
```

```python
region = df.groupby("Region")["Sales"].sum()
print(region)
```

```python
region.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
plt.hist(df["Sales"], bins=15)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(df["Discount"], df["Sales"])

plt.title("Discount vs Sales")
plt.xlabel("Discount")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
monthly_sales = df.groupby(df["Order_Date"].dt.to_period("M"))["Sales"].sum()
plt.figure(figsize=(10,5))
plt.plot( monthly_sales.index.astype(str), monthly_sales.values,marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(7,7))
plt.pie( category_sales, labels=category_sales.index,autopct="%1.1f%%")
plt.title("Sales Contribution by Category")
plt.show()
```
